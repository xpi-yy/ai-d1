## 开始之前：先补一个 20 分钟的小东西

**`guess.py` 猜数字游戏** —— 从 D2 一直欠到现在。

为什么现在补：`while` 循环和 `break` 在 **D14 手写 ReAct Agent 循环**时会直接用到。Agent 的执行循环就是「while 还有任务没完成：想一想 → 调工具 → 看结果 → 再想」。

不补的话，到那天你会卡在「循环怎么写」上面，而不是卡在 Agent 逻辑上。

**做法**：新建 `guess.py`，写一个函数 `play_guess()`，规则：

- 电脑随机想一个 1–100 的数（`import random` → `random.randint(1, 100)`）
- 用户反复输入猜测（`input()`）
- 猜小提示「太小了」，猜大提示「太大了」
- 猜中打印「猜对了！你一共猜了 N 次」然后 `break`

**骨架**（自己补全）：

```
import random

def play_guess():
    答案 = random.randint(1, 100)
    次数 = 0
    print("我已经想好了一个 1 到 100 的数字，你猜猜看！")

    while True:
        猜测 = int(input("请输入你猜的数字："))
        次数 = 次数 + 1

        if 猜测 == 答案:
            print(f"猜对了！你一共猜了 {次数} 次")
            break
        elif 猜测 < 答案:
            print("太小了，再大一点")
        else:
            print("太大了，再小一点")
```

写完在 Jupyter 里跑 `play_guess()` 就能玩。**这一步只是让你亲手用过 `while` + `break`，不要花超过 20 分钟。**

# ④ 对照参考（写完四个问题再看这里）

## 参考表述 1：Self-Attention 在干嘛？

> 句子里的**每个词**，都去看一眼**其他所有词**，算出一个「我该多关注谁」的分数；然后按这个分数，把其他词的信息按比例揉进自己的表示里。
>
> 比如「那只动物没有过马路，因为**它**太累了」——模型在处理「它」这个词时，会重点看「动物」，于是「它」的表示里就带上了「动物」的含义。
>
> **一句话**：让每个词都能看到全句上下文，而不是只看自己邻居。

## 参考表述 2：为什么要「多头」？

> 一组注意力只能关注**一种**关系。
>
> 多头 = 同时开好几组注意力，每组学不同的关注方式：一组看「谁指代谁」、一组看「语法主谓宾」、一组看「语义相似」……
>
> 最后把几组的结果拼起来，信息更全。
>
> **一句话**：一个视角不够用，就多开几个视角，最后汇总。

## 参考表述 3：位置编码解决什么问题？

> **注意力机制本身是不分先后的。** 它把句子里所有词平等地摆在一起看，不知道谁在前、谁在后。
>
> 但语序非常重要：「猫追狗」和「狗追猫」是两件相反的事。
>
> 位置编码就是给每个位置加一个独特的标记（一串有规律的数字），让模型能区分「这是第 3 个词」还是「这是第 7 个词」。
>
> **一句话**：注意力天生不分顺序，得额外告诉它词的先后。

## 参考表述 4：残差连接解决什么问题？

> 网络层数一多，梯度在反向传播时会越传越小（梯度消失），深层根本训不动。
>
> 残差连接就是把**输入原样加回输出**：`输出 = 这一层算出来的结果 + 原来的输入`。
>
> 这样梯度就有一条「直通车道」可以直接传回去，所以模型能堆很多层。
>
> **一句话**：给梯度开一条捷径，让深网络能训得动。
