## D2-1 新语法：while 循环

你学过的 `for` 是「**把列表里的东西挨个拿出来**」，循环次数是固定的。

`while` 是「**只要条件还成立，就一直重复**」，循环次数不固定——猜数字就属于这种：你不知道要猜几次。

看这个例子：从 1 数到 5。

````python
# 用 while 从 1 数到 5
n = 3                   # 起点
while n <= 5:            # 只要 n 还小于等于 5，就继续
    print("现在的 n =", n)
    n = n + 1            # 关键：每圈都要让 n 变大，否则永远循环不完
print("循环结束，n =", n)
````

**读法**：`while 条件:` → 条件成立就执行缩进的代码 → 再回来检查条件 → 成立再执行……直到条件不成立。

**最容易犯的错：忘了让变量变化。** 上面如果把 `n = n + 1` 删掉，`n` 永远是 1，条件永远成立，程序就**死循环**了——你会看到屏幕上疯狂滚动，只能按工具栏的 ⏹ 按钮强制停止。

**改一改**：把 `n = 1` 改成 `n = 3`，看输出有什么变化。

注意最后那行 `print("循环结束...")` 没有缩进——所以它只在循环结束后执行一次。这和 `for` 的规则完全一样。

## D2-3 任务：写一个猜数字游戏

**规则**：电脑随机想一个 1–100 的数。用户反复猜，太小就提示「太小了」，太大就提示「太大了」，猜中就说「猜对了」并告诉用户用了几次。

**完整运行效果**（在终端里）：**

```
我已经想好了一个 1 到 100 的数字，你猜猜看！
请输入你猜的数字：50
太小了，再大一点
请输入你猜的数字：75
太大了，再小一点
请输入你猜的数字：63
猜对了！你一共猜了 3 次
```

**骨架**（在下面新建一个格子写，或者建成 `guess.py` 文件）：

```
import random

def play_guess():
    答案 = random.randint(1, 100)
    次数 = 0
    print("我已经想好了一个 1 到 100 的数字，你猜猜看！")

    while True:                          # 一直循环，直到猜中才退出
        猜测 = int(input("请输入你猜的数字："))
        次数 = 次数 + 1                  # 累加器：记猜了几次

        if 猜测 == 答案:
            print(f"猜对了！你一共猜了 {次数} 次")
            break                        # ← 新东西：break 立即跳出循环
        elif 猜测 < 答案:
            print("太小了，再大一点")
        else:
            print("太大了，再小一点")

play_guess()
```

**里面有两个你没见过的东西：**

| 新东西 | 作用 |
|---|---|
| `while True:` | 条件永远为真，所以会一直循环 |
| `break` | 立刻跳出循环（不然 `while True` 就出不来了） |

**想一想**：为什么用 `while True` 而不是 `while 猜测 != 答案`？（提示：第一次进循环的时候，`猜测` 这个变量还不存在。）

## D2-5 新语法：字典可以直接赋值建键

复习一下你之前数数的写法（D0 作业里用过）：

```
if x in counts:
    counts[x] = counts[x] + 1     # 已经有这个键，加一
else:
    counts[x] = 1                 # 没有这个键，建一个
```

但如果次数能**一次性算出来**（比如用 `count`），就不需要这个 if 了——直接赋值就行：

```
counts[关键词] = 句子.count(关键词)
```

**为什么可以直接写？** 因为 `counts[键] = 值` 这句本身有两个作用：**键不存在就新建，存在就覆盖。** 所以只要你能算出值，一句话就够了。

````python
counts = {}
关键词列表 = ["检索", "幻觉", "向量"]
句子 = "检索很重要，检索做得好，幻觉就少"

for kw in 关键词列表:
    counts[kw] = 句子.count(kw)

print(counts)
````

## D3-1 新语法：读文件

程序要处理数据，第一步永远是**把文件读进来**。

```
文件 = open("sample/技术笔记.txt", "r", encoding="utf-8")

```

| 部分 | 意思 |
|---|---|
| `"sample/技术笔记.txt"` | 文件的路径（相对于 notebook 所在目录） |
| `"r"` | 模式：r = read 读取（还有 "w" 写入） |

| `.close()` | 用完关掉 |

`open / read / close` 每次都要写三行有点烦，所以 Python 提供了更省事的写法：

```
with open("sample/技术笔记.txt", "r", encoding="utf-8") as 文件:
    内容 = 文件.read()
```

`with` 的意思是「这行下面的缩进块用完，自动帮你 close」。**这是推荐写法，以后都用它。**

````python
with open("sample/技术笔记.txt", "r", encoding="utf-8") as 文件:
    内容 = 文件.read()

print("读到了", len(内容.splitlines()), "个行")
print()
print(内容[:120], "……")     # 只打印前 120 个字看看
````

| 写法 | 作用 |

| `内容.splitlines()` | 把文本按行切成一个**列表**（每行一个元素） |
| `len(内容)` | 一共多少个字符 |

**跑出来之后看一眼数字合不合理**——比如「检索」应该出现很多次。这就叫**用数据验证你的代码**。

**进阶（原 D2 计划里的「按次数排序」）**：让结果从多到少排列。

````python
# 排序教学：sorted
计数 = {"检索": 5, "生成": 3, "向量": 2}

# counts.items() 把字典变成一组 (键, 值) 对
print("原始：", list(计数.items()))

# key=lambda 对: 对[1]  的意思是「按每对的第 2 个元素（也就是次数）来比大小」
# reverse=True 意思是「从大到小」
排序后 = sorted(计数.items(), key=lambda 对: 对[1], reverse=True)

print("排序后：", 排序后)
print()
for 词, 次数 in 排序后:
    print(f"  {词}: {次数} 次")
````

**`lambda` 是什么？** 它是一个「一次性的小函数」。这里 `lambda 对: 对[1]` 等价于：

```
def 取次数(对):
    return 对[1]
```

只是省得为这么小的事专门起个名字。**先照抄这个写法，后面你会自然理解。**

## D3-5 关键技能：从嵌套的 JSON 里取出你要的字段

真实 API 返回的结构是**一层套一层**的。看这份数据：

```
{
  "model": "deepseek-chat",
  "choices": [                     ← 这是一个列表！
    {
      "message": {                 ← 列表里的第 0 个元素是个字典
        "content": "..."           ← 我们要的是这个
      }
    }
  ]
}
```

**取法是从外往里一层层剥**：

```
数据["choices"]           → 一个列表
数据["choices"][0]        → 列表的第 0 个元素（一个字典）
数据["choices"][0]["message"]         → 又一层字典
数据["choices"][0]["message"]["content"]   → 终于拿到了
```

**这就是 D4 调 API 时你要做的核心动作。** 写 API 代码，一半时间都在干这件事。

````python
回复 = 数据["choices"][0]["message"]["content"]
print("模型回复：", 回复)
print()

# 顺便把用量也取出来
print("提示 token:", 数据["usage"]["prompt_tokens"])
print("生成 token:", 数据["usage"]["completion_tokens"])
````

### 一个重要提醒：字段可能不存在

真实世界里的 JSON 常常**缺字段**（API 出错时不返回 `choices`，只返回一条错误信息）。这时候直接取就会崩：

```
数据["choices"]     # 如果没有这个键 → KeyError
```

**安全取法**：用 `.get()`，取不到就返回默认值，不会崩。

```
数据.get("choices", [])        # 没有 keys 就返回空列表
数据.get("model", "未知模型")   # 没有就返回 "未知模型"
```

这是 D0 里 `if need in seen:` 的同一个思路——**先确认存在，再取值；或者用 .get 给个兜底。**

````python
# 对比：直接取 vs .get
坏的响应 = {"error": "API 调用超时"}

try:
    print(坏的响应["choices"])
except KeyError as e:
    print("❌ 直接取 → KeyError:", e)

print("✅ 用 .get →", 坏的响应.get("choices", []))
print("✅ 用 .get 给默认值 →", 坏的响应.get("model", "未知模型"))
````

## D3-6 任务：写一个能容错的取字段函数

写一个函数，安全地从 API 响应里取出模型回复的文字。**如果结构不对，返回一句提示，不要崩。**

```
def get_reply(响应):
    # 提示：用 .get() 一层层剥，取不到就返回默认值
    ...
```

自测要求：

```
get_reply(数据)                          → "RAG 的核心是先检索、再生成：检索负责找资料，生成负责根据资料回答。"
get_reply({"error": "超时"})             → "（没有拿到回复）"
```

````python
def get_reply(响应):
    # 提示：choices = 响应.get("choices", [])
    #       如果 choices 是空的，直接返回 "（没有拿到回复）"
    #       否则继续往里面剥
    
    choices =响应.get("choices",[])
    if len(choices)==0:
        return "（没有拿到回复）"
    return choices[0]["message"]["content"]

# 自测
assert get_reply(数据).startswith("RAG 的核心")
assert get_reply({"error": "超时"}) == "（没有拿到回复）"
print("D3-6 通过 ✅")
````
