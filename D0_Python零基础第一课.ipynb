## 2. 变量：给数据起个名字

`age = 25` 你可以理解成：**拿一个盒子，贴上「age」的标签，把 25 放进去。**

以后说「age」，电脑就知道你要的是那个盒子里的东西。

```python
age = 25                 # 把数字 25 装进名叫 age 的盒子
name = "小明"            # 把一串文字装进名叫 name 的盒子

print(name, "今年", age, "岁")
```

注意两点：

1. 文字要用**引号**包起来（`"小明"`），数字不用。
2. `=` 不是数学里的「等于」。它的意思是：**把右边的东西放进左边的盒子。**

```python
age = 18                 # 同一件事再做一次，旧的值被覆盖
print("现在是", age, "岁")
```

**改一改**：把 `age` 改成你自己的年龄。

## 4. 列表：一排编号的格子

列表就是**一排格子，每个格子里放一个东西**。

```python
nums = [2, 7, 11, 15]

print(nums)
print("第一个数是:", nums[0])
print("第二个数是:", nums[1])
print("一共几个数:", len(nums))
```

**新手必踩的第一个坑：编号（下标）从 0 开始，不是从 1。**

所以：2 在第 **0** 格，7 在第 **1** 格，11 在第 2 格，15 在第 3 格。

**改一改**：把 `nums[3]` 打印出来，看看是几。

```python
print(nums[4])
```

**再试一个**：`nums[4]` 会直接报错——因为只有 4 个格子，最大的编号是 3。

报错不用怕。这是电脑在告诉你「你越界了」，它是信息，不是灾难。

## 5. 判断：if

让电脑根据条件做不同的事。

```python
score = 50

if score >= 60:
    print("及格了")
else:
    print("没及格")
```

注意两件事，新手最容易在这翻车：

1. `if score >= 60:` 结尾有个**冒号**
2. 下面那行要**缩进**（前面空几格，通常是 4 格）

缩进不是排版好看，**Python 靠缩进判断「这几行属于 if」**。

**改一改**：把 `score` 改成 50，再运行。

## 6. 循环：让电脑重复做事

如果一个动作要对列表里的每个元素做一遍，就交给 `for`。

```python
nums = [0, 8, 2, 8]

for x in nums:
    print("现在拿到的是:", x)
```

读法是：**「把 nums 里的东西一个一个拿出来，每次装进 x，然后执行下面缩进的那行。」**

这个动作叫**遍历**。它是几乎所有算法题的基础动作。

**改一改**：把列表换成你自己随便想的几个数字。

如果我不只想要**值**，还想要它**在第几格**，就用 `enumerate`：

```python
for i, x in enumerate(nums):
    print("第", i, "格的值是", x)
```

`enumerate` 一次给你两样东西：位置 `i` 和值 `x`。后面那个两数之和的代码里，用的就是这个。

## 7. 把学过的合起来：找出列表里最大的数

**不许用 `max()`**，只用你刚学的：变量 + for + if。

```python
nums = [3, 9, 2, 7]

biggest = nums[0]        # 先假设第 0 个就是最大的
for x in nums:           # 逐个拿出来比
    if x < biggest:      # 如果发现更大的
        biggest = x      # 就把它记下来
print("最大的数是:", biggest)
```

试着读懂这 6 行，然后**动手改成「找最小的数」**。

改法提示：只需要改两个地方——假设哪个是初始值，以及比较的方向。

## 8. 函数：给一段代码起个名字

如果一段代码你要反复用，就给它起个名字，以后叫名字就行。

```python
def add(a, b):
    return a * b


result = add(3, 5)
print("3 * 5 =", result)
```

- `def` 的意思是「我要定义一个函数」
- `add` 是它的名字，括号里的 `a`、`b` 是**入口**（你传什么进去，它就用什么算）
- `return` 是**把结果交出去**

**改一改**：写一个 `multiply(a, b)`，返回 `a * b`，然后算一下 `6 * 7`。

（提示：照抄上面那段，把 `add` 换名字、把 `+` 换成 `*`。）

## 9. 字典：靠名字取东西

列表靠**位置**取（第 0 个），字典靠**名字**取。

```python
person = {"name": "小明", "age": 25,"city": "长沙"}

print(person["name"])
print(person["age"])
print(person["city"])
```

大括号 `{}` 就是字典，里面是一对一对的「**键 : 值**」。

字典有一个了不起的特性：**不管你往里存了多少东西，按「键」去取，几乎都是瞬间完成的。**
这就是算法题里说的「哈希表」——也是让两数之和从「慢」变「快」的关键。

空字典长这样：`seen = {}`

**改一改**：往 `person` 里加一个 `"city"`，然后打印出来。

## 10. 现在回头看昨天那道题

这是「两数之和」的完整代码。**每一行都注释了它对应今天第几节。**

```python
def two_sum(nums, target):
    seen = {}                          # 第 9 节：建一个空字典
    for i, x in enumerate(nums):       # 第 6 节：逐个拿出位置 i 和值 x
        need = target - x              # 第 2 节：算出一个新变量
        if need in seen:               # 第 5 节：判断
            return [seen[need], i]     # 第 8 节：把结果交出去
        seen[x] = i                    # 第 9 节：往字典里存一对
    return []


print(two_sum([2, 7, 11, 15], 9))
```
