## 作业 1：找出第一个重复出现的数字

从左往右看，找出第一个**「之前已经出现过」**的数字。

**例子：**

```
输入 [3, 7, 2, 7, 3]  →  输出 7     （7 的第二次出现，比 3 的第二次出现更早）
输入 [1, 2, 3]        →  输出 None  （一个重复都没有）
输入 [5, 5]           →  输出 5
输入 [9]              →  输出 None
```

**和 `two_sum` 的关系：骨架完全一样**，只是账本记的东西和返回的东西不同。

别忘了：`seen` 是账本，`seen[x] = i` 是登记入册。

````python
def first_duplicate(nums):
    seen = {}
    for i, x in enumerate(nums):
        # ① 想一想：if 什么 in seen ？if x = need
        # ② 想一想：return 什么 ？（数字本身，还是下标？）return(i)
        # ③ 想一想：seen[什么] = 什么 ？sen[i] = x
        if x in seen:
            return x
        seen[x] = i
    return None       # 一个重复都没有（这行不用改）
````

````python
# 作业 1 自测 —— 把上面那个格子填完，再来运行这一格
assert first_duplicate([3, 7, 2, 7, 3]) == 7
assert first_duplicate([1, 2, 3]) is None
assert first_duplicate([5, 5]) == 5
assert first_duplicate([1, 2, 1, 2]) == 1
assert first_duplicate([9]) is None
print("作业 1 全部通过 ✅")
````

## 作业 2（挑战）：找出所有和等于 target 的数对

不是找一对，而是**找出所有配对**，全部放进一个列表返回。

**例子：**

```
输入 [2, 7, 11, 15], 9   →  输出 [[0, 1]]
输入 [1, 2, 3, 4, 5], 6  →  输出 [[1, 3], [0, 4]]
输入 [1, 2], 100         →  输出 []
```

**和 `two_sum` 最大的区别（先想清楚这个再动手）：**

| | 找到一对之后 |
|---|---|
| `two_sum` | 立刻 `return`，任务结束 |
| `two_sum_all` | **继续往下找**，最后一次性返回全部 |

所以你不能再 `return` 了，得换一个动作：**把找到的一对塞进 `result` 这个列表里**。

需要用到的两个新写法（自己动手加进骨架）：

- `result.append(要放进去的东西)` —— 往列表末尾塞一个东西
- 要塞的那个「一对」就是 `[seen[need], i]`

`return result` 要放在**循环外面**（没缩进那里），否则第一圈就返回了。

````python
def two_sum_all(nums, target):
    seen = {}
    result = []                     # 结果列表，已经给你准备好了
    for i, x in enumerate(nums):
        need = target - x
        if need in seen:
            # 想一想：这里不是 return，那应该写什么？
            result.append([seen[need],i])
        seen[x] = i
    return result                   # 循环全部跑完再返回
````

````python
# 作业 2 自测 —— 把上面那个格子填完，再来运行这一格
assert two_sum_all([2, 7, 11, 15], 9) == [[0, 1]]
assert two_sum_all([1, 2, 3, 4, 5], 6) == [[1, 3], [0, 4]]
assert two_sum_all([3, 3], 6) == [[0, 1]]
assert two_sum_all([1, 2], 100) == []
print("作业 2 全部通过 ✅")
````
