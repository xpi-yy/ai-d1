## 先学一个新东西：累加器

作业 3 需要你先理解这个模式。看这段代码：

```
total = 0
for s in [5, 3, 2]:
    total = total + s
print(total)      # 会打印 10
```

**你可能会觉得 `total = total + s` 很奇怪** —— 数学上 `x = x + y` 是不成立的。

因为这里 `=` **不是数学的等于**，是你在第 2 节学过的**赋值**：把右边的结果，放进左边的盒子。

逐圈看就明白了：

| 第几圈 | s | 执行前 total | 执行 total = total + s | 执行后 total |
|---|---|---|---|---|
| 1 | 5 | 0 | 0 + 5 | 5 |
| 2 | 3 | 5 | 5 + 3 | 8 |
| 3 | 2 | 8 | 8 + 2 | 10 |

`total` 就像一个**收纳盒**，每圈往里加一点，循环结束后它就是总和。

**这个模式叫「累加器」，你以后每天都会用到它。** 记住两个固定动作：

- 循环**之前**：`total = 0`（准备一个空盒子）
- 循环**里面**：`total = total + 新的东西`（往里加）

## 作业 3：找出所有高于平均分的分数

**输入**一串分数，**输出**其中所有**严格高于平均分**的分数（保持原来的顺序）。

**例子：**

```
above_average([88, 92, 75, 60])  →  [88, 92]     平均分是 78.75
above_average([100, 100])        →  []           平均分 100，没有比 100 更高的
above_average([50, 100])         →  [100]        平均分 75
above_average([])                →  []           空列表
```

这题要**走两遍列表**：
1. 第一遍：算出总分 → 得到平均分
2. 第二遍：挑出比平均分高的

（为什么不能一遍走完？想一想。）

````python
def above_average(scores):
    if len(scores) == 0:
        return []                       # 空列表没有平均分，直接返回空（这行给你写好了）

    total = 0                           # ← 累加器，从 0 开始
    for s in scores:
        # 想一想：怎么把 s 累加进 total ？
        total = total + s
        

    avg = total / len(scores)           # 平均分算好了，直接用

    result = []
    for s in scores:
        # 想一想：什么条件说明「这个分数高于平均分」？
        if s > avg:
        # 想一想：满足条件时，把什么塞进 result ？
            result.append(s)
        

    return result
````

````python
# 作业 3 自测
assert above_average([88, 92, 75, 60]) == [88, 92]
assert above_average([100, 100]) == []
assert above_average([50, 100]) == [100]
assert above_average([]) == []
print("作业 3 全部通过 ✅")
````

## 作业 4：关键词命中打分

给一段**文本**和一组**关键词**，数出这段文本里命中了几个关键词。

**例子：**

```
输入 "RAG 的检索优化与幻觉抑制", 关键词 ["RAG", "检索", "微调"]
输出 2            （命中了 RAG、检索；没命中 微调）

输入 "今天天气很好", 关键词 ["RAG", "检索"]
输出 0

输入 "微调 微调 微调", 关键词 ["微调"]
输出 1            （每个关键词最多算一次，出现三遍也只算 1）
```

**需要的只有一个新写法**：`关键词 in 文本` —— Python 里 `in` 也能用来判断「一小段文字是不是出现在另一段文字里」，结果同样是 `True` 或 `False`。

（你对 `in` 已经不陌生了：`need in seen` 是查字典，`x in nums` 是查列表，这次是查文本。）

累加器和作业 3 完全一样。

````python
def count_hits(text, keywords):
    hits = 0                              # 累加器（和作业 3 同一个模式）
    for kw in keywords:
        # 想一想：怎么判断「关键词 kw 出现在文本 text 里」？
        if kw in text:
        # 命中就把 hits 加 1
            hits = hits + 1
       
    return hits
````

````python
# 作业 4 自测
assert count_hits("RAG 的检索优化与幻觉抑制", ["RAG", "检索", "微调"]) == 2
assert count_hits("今天天气很好", ["RAG", "检索"]) == 0
assert count_hits("微调 微调 微调", ["微调"]) == 1
assert count_hits("", ["RAG"]) == 0
print("作业 4 全部通过 ✅")
````

## 加餐（选做，但它才是重点）

给一个问题（几个关键词）和一堆文档，返回**得分最高**的那篇文档。

**例子：**

```

返回 "RAG 检索优化"     （分数相同时，返回先出现的那篇）
```

**它和你在 D0 第 7 节写的「找最大的数」是同一个套路**：都是「边走边记住当前最好的那个」。

区别只有一个：第 7 节记住的是**数字**，这次要记住的是**一整篇文档**。

```
第 7 节：biggest = nums[0]  →  biggest = x
这  次：best = None         →  best = doc
```

> 提示：`best_score` 我给了 `-1` 作为初值。原因是——任何文档的得分都 >= 0，所以 -1 保证第一轮一定会被替换掉。你可以想想如果不给 -1 会怎样。

````python
def best_doc(question_keywords, docs):
    best = None
    best_score = -1                       # 为什么从 -1 开始？想想上面那段提示
    for doc in docs:
        score = count_hits(doc, question_keywords)
        if score > best_score:
            best_score = score            # 记住新的最高分
            # 想一想：这里还应该记住什么？一行代码
            pass
    return best
````

````python
# 加餐自测（需要作业 4 已经做对）
docs = ["RAG 检索优化", "微调模型的方法", "RAG 与检索的结合"]
assert best_doc(["RAG", "检索"], docs) == "RAG 检索优化"
assert best_doc(["微调"], docs) == "微调模型的方法"
assert best_doc(["找不到的词"], docs) == "RAG 检索优化"   # 全部 0 分时返回第一篇
print("加餐 全部通过 ✅")
````
